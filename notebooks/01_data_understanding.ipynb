```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display all columns when printing DataFrames
pd.set_option("display.max_columns", None)

# Display more rows before truncating the output
pd.set_option("display.max_rows", 100)
```

```python
# Load customer information
customers = pd.read_csv("../data/raw/olist_customers_dataset.csv")

# Load order information
orders = pd.read_csv("../data/raw/olist_orders_dataset.csv")

# Load order item details
order_items = pd.read_csv("../data/raw/olist_order_items_dataset.csv")

# Load payment information
payments = pd.read_csv("../data/raw/olist_order_payments_dataset.csv")

# Load customer reviews
reviews = pd.read_csv("../data/raw/olist_order_reviews_dataset.csv")

# Load product information
products = pd.read_csv("../data/raw/olist_products_dataset.csv")

# Load seller information
sellers = pd.read_csv("../data/raw/olist_sellers_dataset.csv")

# Load geolocation information
geolocation = pd.read_csv("../data/raw/olist_geolocation_dataset.csv")

# Load product category translations
category_translation = pd.read_csv("../data/raw/product_category_name_translation.csv")
```

```python
# Store every DataFrame in one dictionary.
# This lets us loop through all datasets instead of
# writing the same code repeatedly.

datasets = {
    "customers": customers,
    "orders": orders,
    "order_items": order_items,
    "payments": payments,
    "reviews": reviews,
    "products": products,
    "sellers": sellers,
    "geolocation": geolocation,
    "category_translation": category_translation,
}
```

```python
for name, df in datasets.items():

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    # Number of rows and columns
    print(f"Shape: {df.shape}")

    # Column names
    print("\nColumns:")
    print(df.columns.tolist())

    # Data types
    print("\nData Types:")
    print(df.dtypes)

    # Missing values
    print("\nMissing Values:")
    print(df.isnull().sum())

    # Duplicate rows
    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    # First five records
    print("\nFirst Five Rows:")
    display(df.head())
```

**4. Dataset Understanding**

In this section, we will examine each dataset to understand its purpose, the information it contains, its key columns, its role in the business process, and how it connects with other datasets.

**Customers Dataset - Observations**

- The dataset contains **99,441 customer records** and **5 columns**.
- Each record represents customer information associated with an order.
- The table includes customer IDs and location details (ZIP code, city, and state).
- No missing values were found in any column.
- The data types are appropriate for the stored information.
- This table will be used to identify customers and analyze customer distribution by location.

**Orders Dataset - Observations**

- The dataset contains **99,441 orders** and **8 columns**.
- Each row represents a single customer order.
- The `order_id` column uniquely identifies each order.
- The `customer_id` column links each order to a customer.
- Several columns store timestamps representing different stages of the order lifecycle.
- Timestamp columns are currently stored as text and will be converted to datetime during data cleaning.
- Missing values are present in delivery-related columns, which may be expected for cancelled or undelivered orders.
- This is the central table of the project and will be used for customer behavior, delivery performance, and churn analysis.

**Order Items Dataset - Observations**

- The dataset contains **112,650 order items** and **7 columns**.
- Each row represents a single product within an order.
- A single order can contain multiple products, resulting in multiple rows with the same `order_id`.
- The table connects orders, products, and sellers using their respective IDs.
- The `price` and `freight_value` columns will be used to calculate revenue and shipping costs.
- The `shipping_limit_date` column should be converted to a datetime format during data cleaning.
- No missing values were found in this dataset.

**Payments Dataset - Observations**

- The dataset contains **103,886 payment records** and **5 columns**.
- Each row represents a payment transaction associated with an order.
- A single order can have multiple payment transactions, resulting in multiple rows with the same `order_id`.
- The dataset includes payment method, installment count, and payment amount.
- The data types are appropriate for analysis.
- No missing values were found in this dataset.
- This table will be used to analyze payment preferences, installment usage, and revenue.

**Reviews Dataset - Observations**

- The dataset contains **99,224 customer reviews** and **7 columns**.
- Each review is associated with an order through `order_id`.
- The `review_score` column measures customer satisfaction on a scale of 1 to 5.
- A large number of missing values exist in `review_comment_title` and `review_comment_message`, which is expected because many customers provide only a rating without written feedback.
- Date columns will be converted to datetime format during data cleaning.
- This table will be used to analyze customer satisfaction and its relationship with delivery performance and purchasing behavior.

**Products Dataset - Observations**

- The dataset contains **32,951 unique products** and **9 columns**.
- Each row represents a unique product sold on the platform.
- The dataset includes product categories, descriptions, images, and physical dimensions.
- A small number of missing values exist in category and product metadata columns.
- Product dimensions and weight will support logistics and shipping analysis.
- This table will be joined with `order_items` to analyze product sales and category performance.

**Sellers Dataset - Observations**

- The dataset contains **3,095 sellers** and **4 columns**.
- Each row represents one unique seller on the marketplace.
- The `seller_id` serves as the primary key and links this table with the `order_items` table.
- Seller city and state information support regional performance analysis.
- No missing values were found in this dataset, indicating good data quality.

**Geolocation Dataset - Observations**

- The dataset contains **1,000,163 geographic records** and **5 columns**.
- It maps ZIP code prefixes to latitude, longitude, city, and state.
- Multiple records can share the same ZIP code prefix because they represent different geographic coordinates within the same postal region.
- The dataset contains no missing values.
- This table supports geographic analysis, logistics optimization, and regional business intelligence.

**Product Category Translation Dataset - Observations**

- The dataset contains **71 product category translations** and **2 columns**.
- It maps Portuguese product category names to their English equivalents.
- This is a lookup table and does not contain transactional business data.
- It will be joined with the `products` table to improve report readability.
- No missing values were found in this dataset.
