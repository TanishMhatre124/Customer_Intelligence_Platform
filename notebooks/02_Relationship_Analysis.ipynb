````python
# Import pandas for data analysis
import pandas as pd


# Load datasets
customers = pd.read_csv("../data/raw/olist_customers_dataset.csv")
orders = pd.read_csv("../data/raw/olist_orders_dataset.csv")
order_items = pd.read_csv("../data/raw/olist_order_items_dataset.csv")
payments = pd.read_csv("../data/raw/olist_order_payments_dataset.csv")
reviews = pd.read_csv("../data/raw/olist_order_reviews_dataset.csv")
products = pd.read_csv("../data/raw/olist_products_dataset.csv")
sellers = pd.read_csv("../data/raw/olist_sellers_dataset.csv")


# Store datasets together for easy checking
datasets = {
    "customers": customers,
    "orders": orders,
    "order_items": order_items,
    "payments": payments,
    "reviews": reviews,
    "products": products,
    "sellers": sellers
}


# Check dataset size
for name, df in datasets.items():
    print(name, df.shape)
````

**2. Identify Primary Keys**

Check whether important ID columns contain duplicate values.
A primary key should uniquely identify each row.

````python
# Check uniqueness of primary key candidates

key_checks = {
    "customers": "customer_id",
    "orders": "order_id",
    "products": "product_id",
    "sellers": "seller_id",
    "reviews": "review_id"
}


for table, key in key_checks.items():
    
    # Count duplicate IDs
    duplicates = datasets[table][key].duplicated().sum()
    
    print(f"{table} - {key}")
    print("Duplicate values:", duplicates)
    print("-" * 40)
````

**3. Validate Foreign Keys**

Check whether foreign key values correctly match with primary keys in related tables.

 General Pattern: Validate Foreign Keys

```python
missing_records = child_table[
    ~child_table["foreign_key"].isin(parent_table["primary_key"])
]
```

````python
# Check orders with missing customers

missing_customers = orders[
    ~orders["customer_id"].isin(customers["customer_id"])
]

print("Orders without matching customers:", len(missing_customers))
````

**Orders → Order Items Relationship**

Check whether every order item is connected to a valid order.

````python
# Check order items without matching orders

missing_orders = order_items[
    ~order_items["order_id"].isin(orders["order_id"])
]

print("Order items without matching orders:", len(missing_orders))
````

**Products → Order Items Relationship**

Check whether every product in an order exists in the products table.

````python
# Check order items without matching products

missing_products = order_items[
    ~order_items["product_id"].isin(products["product_id"])
]

print("Order items without matching products:", len(missing_products))
````

**Sellers → Order Items Relationship**

Check whether every seller in an order exists in the sellers table.

````python
# Check order items without matching sellers

missing_sellers = order_items[
    ~order_items["seller_id"].isin(sellers["seller_id"])
]

print("Order items without matching sellers:", len(missing_sellers))
````

**4. Data Model Understanding**

Identify the fact table, dimension tables, and understand how the datasets are connected.

### Fact Table

**Orders**

The `orders` table is the fact table because it records every customer purchase and connects with multiple related tables.

### Dimension Tables

- Customers
- Products
- Sellers
- Payments
- Reviews
- Category Translation

These tables provide additional information about orders.

### Database Relationships

```text
Customers
    │
customer_id
    │
    ▼
Orders
    │
    ├──────────────┐
    ▼              ▼
Order Items     Payments
    │              │
    ▼              ▼
Products       Reviews
    │
    ▼
Category Translation

Order Items
    │
    ▼
Sellers
```

### Key Observations

- The `orders` table is the central table.
- Every order belongs to a customer.
- Every order item belongs to an order.
- Every product sold exists in the products table.
- Every seller exists in the sellers table.
- All key relationships were successfully validated.
